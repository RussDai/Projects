# **pd.melt():**

`pandas.melt()` is used to **transform a wide-format DataFrame into a long-format DataFrame**. This means it takes columns and **turns them into rows**, making our data *tidy* for analysis or plotting.

**`pd.melt()` is Useful when:**

* we have **one or more identifier columns** and the rest are **measured variables**.

* we want to reshape our DataFrame for:
  * Easier `plotting` (e.g., with seaborn or matplotlib).
  * `GroupBy` operations.
  * Exporting tidy data.
  * Feeding into ML models or APIs expecting long-form data.

```python
pd.melt(
    frame= data,         # data is our dataframe
    id_vars=None,        # columns to keep fixed (identifiers)
    value_vars=None,     # columns to unpivot (default: all columns not in id_vars)
    var_name=None,       # name for new 'variable' column
    value_name='value'   # name for new 'value' column
)
```

**Example 1: Basic usage:**

````python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Alice', 'Bob'],
    'Math': [90, 80],
    'Science': [85, 75]
})
df
````

````python
melted = pd.melt(frame= df,  # our dataframe
                 id_vars='Name',  # Column to keep fixed
                 var_name='Subject', # name for new variable column
                 value_name='Score') # name for new value colmn
melted
````

Now it's ready for `groupby`, `aggregation`, or `plotting`.

**Example 2: Melting with multiple `id\_vars`:**

````python
df = pd.DataFrame({
    'Student': ['A', 'B'],
    'Year': [2023, 2023],
    'Math': [90, 80],
    'Science': [88, 78]
})
df
````

````python
melted = pd.melt(df, # dataframe 
    id_vars=['Student', 'Year'],  # columns to keep fixed 
    var_name='Subject', # name for new variable column
    value_name='Marks')  # name for new value column

melted
````

**Example 3: When all columns are melted (no id\_vars):**

````python
df = pd.DataFrame({
    'Jan': [100, 200],
    'Feb': [150, 250]
})
df
````

````python
melted = pd.melt(df, # dataframe 
    var_name='Month', # name for new variable column
    value_name='Sales') # name for new value colmn
melted
````

### **opposite** of `pd.melt()`?

The opposite operation is **`pd.pivot()`** or **`pd.pivot_table()`** (if we need aggregation).

````python
# Original wide-format DataFrame
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Math': [85, 78, 92],
    'Science': [90, 82, 87],
    'English': [88, 76, 85]
})

print("Original wide-format DataFrame:")
print(df)
print("="*50)
# Melt the DataFrame (long format)
melted = pd.melt(df, id_vars='Name', var_name='Subject', value_name='Score')

print("\nMelted (long) DataFrame:")
print(melted)
print("="*50)
# Reverse the melt operation using pivot
wide = melted.pivot(index='Name', columns='Subject', values='Score').reset_index()

print("\nReversed (wide) DataFrame:")
print(wide)
````

> 🔁 `pd.melt()` → "long format"  
> 
> 🔁 `pd.pivot()` → "wide format"

**Example 4: Full melt-pivot round trip:**

````python
original = pd.DataFrame({
    'Person': ['Tom', 'Jerry'],
    'Jan': [20, 25],
    'Feb': [21, 23]
})
original
````

````python
#Melt
long_df = pd.melt(original, id_vars='Person', var_name='Month', value_name='Temp')
long_df
````

````python
# Pivot (to recover original)
wide_df = long_df.pivot(index='Person', columns='Month', values='Temp').reset_index()
wide_df
````

**When Not to Use `melt()`:**

* When our data is already in long form.
* If our plan is to do wide-format matrix operations (e.g., correlation matrices, pivot heatmaps).
