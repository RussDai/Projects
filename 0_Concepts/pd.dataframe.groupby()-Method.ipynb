# **groupby():**

The `pandas.DataFrame.groupby()` method is used to **split** data into groups based on some criteria, **`apply`** a function to each group independently, and then **`combine`** the results back into a DataFrame or Series.

1. **`Split`** the data into groups.
2. **`Apply`** a function to each group (e.g., `sum`, `mean`).
3. **`Combine`** the results into a new DataFrame or Series.

When we call `df.groupby(...)`, it returns a **GroupBy object**.
This object is essentially a **view of the original DataFrame grouped by some key(s)** (like column values).

No computation happens until you call an aggregation or transformation method (e.g., `.sum()`, `.mean()`, etc.).

```python
import pandas as pd

data = {
    'Department': ['HR', 'HR', 'IT', 'IT', 'Finance'],
    'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Salary': [50000, 52000, 60000, 62000, 58000]
}
df = pd.DataFrame(data)
df
```

```python
grouped = df.groupby('Department')
grouped
```

Now `grouped` is a **GroupBy object**. Nothing has been computed yet.

### **Aggregation Methods (trigger computation):**

| Method              | Description                       |
| ------------------- | --------------------------------- |
| `.sum()`            | Sum of each group                 |
| `.mean()`           | Mean of each group                |
| `.median()`         | Median                            |
| `.count()`          | Count of non-NA entries per group |
| `.size()`           | Number of rows per group          |
| `.min()` / `.max()` | Min / Max of each group           |
| `.std()` / `.var()` | Standard deviation / Variance     |

```python
grouped['Salary'].sum()  # total salary per department
```

### **Transformation / Filtering:**

| Method         | Description                                                  |
| -------------- | ------------------------------------------------------------ |
| `.transform()` | Returns object of same shape; useful for feature engineering |
| `.filter()`    | Filters groups based on a condition                          |
| `.apply()`     | Apply a custom function to each group                        |
| `.agg()`       | Apply multiple aggregation functions                         |

```python
# Add a column showing salary mean of each department
df['Dept_Avg_Salary'] = grouped['Salary'].transform('mean')
df
```

### **Iterating over groups:**

```python
for name, group in grouped:
    print(f"Group: {name}")
    print(group)
```

## **GroupBy Object Attributes:**

| Attribute  | Description                                 |
| ---------- | ------------------------------------------- |
| `.groups`  | Dictionary mapping group name → row indices |
| `.ngroups` | Number of groups                            |
| `.keys`    | Keys used for grouping                      |
| `.dtypes`  | Data types of grouped data columns          |

```python
grouped.groups    # {'Finance': [4], 'HR': [0, 1], 'IT': [2, 3]}
```

```python
grouped.ngroups   # 3
```

## **Grouping by Multiple Columns:**

```python
df.groupby(['Department', 'Employee']).sum()
```


---

## 📝 Summary

* `groupby()` returns a **lazy GroupBy object**.
* You **trigger computation** using methods like `.sum()`, `.mean()`, `.agg()`, `.transform()`.
* Use `.groups`, `.ngroups`, and iteration to inspect the GroupBy object.
* Combine with custom functions for powerful data analysis workflows.
